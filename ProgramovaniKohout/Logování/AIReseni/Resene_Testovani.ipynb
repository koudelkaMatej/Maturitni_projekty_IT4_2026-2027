# Modul `unittest` — základní struktura

Python má vestavěný modul `unittest` — **nemusíme nic instalovat**.

## Jak napsat test?

1. Importujeme `unittest`
2. Vytvoříme třídu, která **dědí** z `unittest.TestCase`
3. Každá testovací metoda **začíná na `test_`**
4. Uvnitř metody používáme **assertovací metody** (`assertEqual`, `assertTrue`, …)
5. Spustíme testy pomocí `unittest.main()`

## Syntaxe

```python
import unittest

class TestMojeFunkce(unittest.TestCase):
    def test_pripad_1(self):
        self.assertEqual(funkce(vstup), ocekavany_vysledek)

    def test_pripad_2(self):
        self.assertTrue(funkce(vstup))

if __name__ == '__main__':
    unittest.main()
```

> **⚠️ Pozor:** V Jupyter Notebooku `unittest.main()` nefunguje správně, protože volá `sys.exit()`. Proto v notebooku vytváříme testovací sadu ručně (viz příklady níže). **Doporučujeme testovat v `.py` souborech.**

### Příklad — první test

````python
import unittest

# Definice funkce, kterou chceme testovat
def secti(x, y):
    return x + y

# Testovací třída — dědí z unittest.TestCase
class TestSectiFunkce(unittest.TestCase):
    def test_kladna_cisla(self):
        self.assertEqual(secti(2, 3), 5)

    def test_zaporna_cisla(self):
        self.assertEqual(secti(-1, 1), 0)

    def test_nuly(self):
        self.assertEqual(secti(0, 0), 0)

# Spuštění testů v Jupyter Notebooku (náhrada za unittest.main())
suite = unittest.TestLoader().loadTestsFromTestCase(TestSectiFunkce)
runner = unittest.TextTestRunner(verbosity=2)
runner.run(suite)
````

> **Tip:** `verbosity=2` zajistí podrobný výpis — uvidíme název každého testu a jeho výsledek (`ok` / `FAIL`).

> **Tip:** `unittest` automaticky spustí **všechny metody**, které začínají na `test_`. Metody bez tohoto prefixu se ignorují.

# Assertovací metody — přehled

Assertovací metody jsou **srdcem testování**. Každá z nich ověřuje jinou podmínku. Pokud podmínka neplatí, test **selže** a zobrazí se chybová zpráva.

## Nejčastější assertovací metody

| Metoda | Ověřuje | Příklad |
|---|---|---|
| `assertEqual(a, b)` | `a == b` | `assertEqual(secti(2, 3), 5)` |
| `assertNotEqual(a, b)` | `a != b` | `assertNotEqual(secti(2, 3), 6)` |
| `assertTrue(x)` | `x is True` | `assertTrue(je_sude(4))` |
| `assertFalse(x)` | `x is False` | `assertFalse(je_sude(3))` |
| `assertIs(a, b)` | `a is b` (stejný objekt) | `assertIs(x, None)` |
| `assertIsNone(x)` | `x is None` | `assertIsNone(vysledek)` |
| `assertIn(a, b)` | `a in b` | `assertIn("a", "ahoj")` |
| `assertNotIn(a, b)` | `a not in b` | `assertNotIn("x", "ahoj")` |
| `assertRaises(Error)` | Vyvolá se výjimka | `assertRaises(ValueError)` |
| `assertAlmostEqual(a, b)` | `a ≈ b` (pro desetinná čísla) | `assertAlmostEqual(0.1 + 0.2, 0.3)` |

> **Pozn.:** Všechny metody přijímají volitelný parametr `msg` pro vlastní chybovou zprávu: `assertEqual(a, b, msg="Hodnoty se neshodují!")`

### Příklad — ukázka různých assertů

````python
import unittest

def je_sude(n):
    return n % 2 == 0

def deleni(a, b):
    if b == 0:
        raise ValueError("Nelze dělit nulou!")
    return a / b

class TestRuzneAsserty(unittest.TestCase):
    def test_assertEqual(self):
        self.assertEqual(2 + 2, 4)

    def test_assertNotEqual(self):
        self.assertNotEqual(2 + 2, 5)

    def test_assertTrue(self):
        self.assertTrue(je_sude(4))

    def test_assertFalse(self):
        self.assertFalse(je_sude(3))

    def test_assertIn(self):
        self.assertIn("a", "ahoj")

    def test_assertRaises(self):
        with self.assertRaises(ValueError):
            deleni(10, 0)

suite = unittest.TestLoader().loadTestsFromTestCase(TestRuzneAsserty)
runner = unittest.TextTestRunner(verbosity=2)
runner.run(suite)
````

> **Tip:** `assertRaises` se používá s kontextovým manažerem `with`. Kód uvnitř bloku `with` by měl vyvolat uvedenou výjimku — pokud se tak nestane, test selže.

# Co se stane, když test selže?

Pokud assertovací metoda zjistí nesoulad, test **selže** (FAIL). Unittest vypíše:
- **Který test selhal** (název metody)
- **Co se očekávalo** vs. **co se skutečně stalo**
- **Traceback** s přesným řádkem

### Příklad — neúspěšný test

````python
import unittest

def secti(x, y):
    return x + y

class TestSectiFunkce(unittest.TestCase):
    def test_spravny(self):
        self.assertEqual(secti(2, 3), 5)  # ✅ projde

    def test_chybny(self):
        self.assertEqual(secti(2, 3), 6)  # ❌ selže — 2+3 = 5, ne 6!

suite = unittest.TestLoader().loadTestsFromTestCase(TestSectiFunkce)
runner = unittest.TextTestRunner(verbosity=2)
runner.run(suite)
````

> **Tip:** Ve výstupu uvidíme `FAIL` u `test_chybny` a podrobnosti: `AssertionError: 5 != 6`. To nám přesně řekne, co bylo očekáváno a co funkce skutečně vrátila.

# Testování hranic vstupů

Důležité je testovat nejen „normální" vstupy, ale i **krajní případy** (boundary testing):
- Co když je vstup **nula**?
- Co když je vstup **záporný**?
- Co když je vstup **na hranici** povoleného rozsahu?

### Příklad — funkce omezující číslo do rozsahu

# Testování složitějších datových struktur

Lze otestovat, zda funkce pracuje správně se **seznamy**, **slovníky** či **n-ticemi**. Assertovací metody fungují i pro porovnávání celých kolekcí.

### Příklad — součet odpovídajících prvků dvou seznamů

````python
import unittest

def soucty_prvku(list1, list2):
    """Sečte odpovídající prvky dvou seznamů stejné délky."""
    if len(list1) != len(list2):
        raise ValueError("Seznamy musí mít stejnou délku!")
    return [a + b for a, b in zip(list1, list2)]

class TestSouctyPrvku(unittest.TestCase):
    def test_spravne_scitani(self):
        self.assertEqual(soucty_prvku([1, 2, 3], [4, 5, 6]), [5, 7, 9])

    def test_prazdne_seznamy(self):
        self.assertEqual(soucty_prvku([], []), [])

    def test_nerovnaka_delka(self):
        with self.assertRaises(ValueError):
            soucty_prvku([1, 2], [1, 2, 3])

suite = unittest.TestLoader().loadTestsFromTestCase(TestSouctyPrvku)
runner = unittest.TextTestRunner(verbosity=2)
runner.run(suite)
````

# Testování stavového chování objektů

U objektově orientovaného kódu testujeme, zda objekty **správně mění svůj stav** po volání metod. To je důležité zejména u tříd, které si uchovávají vnitřní data.

### Příklad — třída BankovniUcet

````python
import unittest

class BankovniUcet:
    def __init__(self, pocatecni_zustatek=0):
        self.zustatek = pocatecni_zustatek

    def vklad(self, castka):
        self.zustatek += castka

    def vyber(self, castka):
        if castka > self.zustatek:
            raise ValueError("Nedostatečné prostředky!")
        self.zustatek -= castka

class TestBankovniUcet(unittest.TestCase):
    def test_vklad(self):
        ucet = BankovniUcet(100)
        ucet.vklad(50)
        self.assertEqual(ucet.zustatek, 150)

    def test_vyber(self):
        ucet = BankovniUcet(100)
        ucet.vyber(30)
        self.assertEqual(ucet.zustatek, 70)

    def test_nedostatecny_zustatek(self):
        ucet = BankovniUcet(50)
        with self.assertRaises(ValueError):
            ucet.vyber(100)

    def test_pocatecni_zustatek(self):
        ucet = BankovniUcet()
        self.assertEqual(ucet.zustatek, 0)

suite = unittest.TestLoader().loadTestsFromTestCase(TestBankovniUcet)
runner = unittest.TextTestRunner(verbosity=2)
runner.run(suite)
````

# 📋 Shrnutí

| Koncept | Co dělá | Příkaz |
|---|---|---|
| Import | Načte testovací modul | `import unittest` |
| Testovací třída | Seskupuje testy | `class TestX(unittest.TestCase):` |
| Testovací metoda | Jeden test (prefix `test_`) | `def test_pripad(self):` |
| Porovnání | Ověří rovnost | `self.assertEqual(a, b)` |
| Pravdivost | Ověří `True` / `False` | `self.assertTrue(x)` / `self.assertFalse(x)` |
| Výjimka | Ověří vyvolání chyby | `with self.assertRaises(Error):` |
| Obsaženost | Ověří přítomnost prvku | `self.assertIn(a, b)` |
| Spuštění | Spustí všechny testy | `unittest.main()` |

### Kdy použít kterou metodu?

| Situace | Metoda |
|---|---|
| Porovnávám dva výsledky | `assertEqual` |
| Ověřuji, zda funkce vrací `True`/`False` | `assertTrue` / `assertFalse` |
| Funkce má vyhodit chybu při špatném vstupu | `assertRaises` |
| Hledám prvek v seznamu/textu | `assertIn` |
| Porovnávám desetinná čísla | `assertAlmostEqual` |

> 🎯 **Tip na závěr:** Pište testy **krátké a zaměřené** — každá testovací metoda by měla ověřovat **jednu konkrétní věc**. Název metody by měl jasně říkat, co testuje.

````python
import unittest

def text_na_velka(text):
    """Převede text na velká písmena a odmítne jiné typy."""
    if not isinstance(text, str):
        raise TypeError("Vstup musí být řetězec (str)")
    return text.upper()  # .upper() funguje také s českou diakritikou.

class TestTextNaVelka(unittest.TestCase):
    def test_mala_pismena(self):
        self.assertEqual(text_na_velka("ahoj"), "AHOJ")

    def test_uz_velka(self):
        self.assertEqual(text_na_velka("PYTHON"), "PYTHON")

    def test_smiseny_text(self):
        self.assertEqual(text_na_velka("AhOj Česko"), "AHOJ ČESKO")

    def test_cislo_vyvola_chybu(self):
        # assertRaises ověřuje, že zavolání opravdu vyvolá TypeError.
        with self.assertRaises(TypeError):
            text_na_velka(123)

suite = unittest.TestLoader().loadTestsFromTestCase(TestTextNaVelka)
vysledek = unittest.TextTestRunner(verbosity=2).run(suite)
assert vysledek.wasSuccessful()
````

## Cvičení 3 — Testování teploměru

Třída `Teplomer` je pro vás již vytvořena (viz kód níže). Vytvořte třídu `TestTeplomer` s následujícími testy:

1. **Test inicializace:** Ověřte, že se teplota správně inicializuje na 0 a maximální teplota na zadanou hodnotu (výchozí je 100).
2. **Test nastavení teploty:** Nastavte teplotu na hodnotu v povoleném rozmezí a ověřte správnost. Zkontrolujte, že pokus nastavit teplotu nad limit vyvolá `ValueError`.
3. **Test zvýšení teploty:** Ověřte, že zvýšení teploty v rámci limitu proběhne správně. Zkontrolujte, že zvýšení nad limit vyvolá `ValueError`.
4. **Test snížení teploty:** Ověřte, že snížení teploty proběhne správně a nedostane se pod −273.15 °C. Pokud je požadavek snížení pod −273.15 °C, ověřte, že se teplota nastaví na −273.15 °C.

<details>
<summary>🔎 Očekávaný výstup (klikni pro zobrazení)</summary>


</details>

````python
import unittest

class Teplomer:
    def __init__(self, max_teplota=100):
        """Inicializuje objekt teploměru s výchozí teplotou 0 a maximální teplotou."""
        self.teplota = 0
        self.max_teplota = max_teplota

    def nastav_teplotu(self, nova_teplota):
        """Nastaví aktuální teplotu. Pokud přesáhne max, vyvolá výjimku."""
        if nova_teplota > self.max_teplota:
            raise ValueError("Teplota přesáhla maximální povolenou hodnotu!")
        self.teplota = nova_teplota

    def oteplit(self, stupne):
        """Zvýší teplotu o zadaný počet stupňů. Pokud přesáhne max, vyvolá výjimku."""
        if self.teplota + stupne > self.max_teplota:
            raise ValueError("Teplota přesáhla maximální povolenou hodnotu!")
        self.teplota += stupne

    def ochladit(self, stupne):
        """Sníží teplotu o zadaný počet stupňů. Minimum je -273.15 °C."""
        nova_teplota = self.teplota - stupne
        self.teplota = max(nova_teplota, -273.15)



# Třída Teplomer je definovaná výše v původním zadání.
# Testy ji pouze používají; nepřepisujeme její implementaci.
class TestTeplomer(unittest.TestCase):
    def test_inicializace(self):
        teplomer = Teplomer()
        self.assertEqual(teplomer.teplota, 0)
        self.assertEqual(teplomer.max_teplota, 100)
        self.assertEqual(Teplomer(75).max_teplota, 75)

    def test_nastav_teplotu(self):
        teplomer = Teplomer()
        teplomer.nastav_teplotu(25)
        self.assertEqual(teplomer.teplota, 25)

    def test_nastav_teplotu_nad_limit(self):
        teplomer = Teplomer()
        with self.assertRaises(ValueError):
            teplomer.nastav_teplotu(101)

    def test_oteplit(self):
        teplomer = Teplomer()
        teplomer.oteplit(30)
        self.assertEqual(teplomer.teplota, 30)

    def test_oteplit_nad_limit(self):
        teplomer = Teplomer()
        teplomer.nastav_teplotu(90)
        with self.assertRaises(ValueError):
            teplomer.oteplit(20)

    def test_ochladit(self):
        teplomer = Teplomer()
        teplomer.ochladit(20)
        self.assertEqual(teplomer.teplota, -20)

    def test_ochladit_pod_absolutni_nulu(self):
        teplomer = Teplomer()
        teplomer.ochladit(500)
        self.assertAlmostEqual(teplomer.teplota, -273.15)

suite = unittest.TestLoader().loadTestsFromTestCase(TestTeplomer)
vysledek = unittest.TextTestRunner(verbosity=2).run(suite)
assert vysledek.wasSuccessful()
````

**Co si z cvičení 3 zapamatovat:** Testujeme běžné hodnoty i hraniční situace, zvlášť maximální teplotu a absolutní nulu.

````python
import unittest

class Kalkulacka:
    def secti(self, a, b):
        return a + b

    def odecti(self, a, b):
        return a - b

    def vynasob(self, a, b):
        return a * b

    def vydel(self, a, b):
        if b == 0:
            raise ValueError("Nulou nelze dělit")
        return a / b

class TestKalkulacka(unittest.TestCase):
    def setUp(self):
        # setUp() se provede před KAŽDÝM testem.
        self.k = Kalkulacka()

    def test_scitani(self):
        self.assertEqual(self.k.secti(2, 3), 5)

    def test_odcitani(self):
        self.assertEqual(self.k.odecti(9, 4), 5)

    def test_nasobeni(self):
        self.assertEqual(self.k.vynasob(3, 4), 12)

    def test_deleni(self):
        self.assertEqual(self.k.vydel(12, 3), 4)

    def test_zaporna_cisla(self):
        self.assertEqual(self.k.secti(-3, 1), -2)
        self.assertEqual(self.k.odecti(-3, 2), -5)
        self.assertEqual(self.k.vynasob(-2, 3), -6)
        self.assertEqual(self.k.vydel(-8, 2), -4)

    def test_nula(self):
        self.assertEqual(self.k.secti(0, 3), 3)
        self.assertEqual(self.k.odecti(0, 3), -3)
        self.assertEqual(self.k.vynasob(0, 3), 0)
        self.assertEqual(self.k.vydel(0, 3), 0)

    def test_deleni_nulou(self):
        with self.assertRaises(ValueError):
            self.k.vydel(10, 0)

suite = unittest.TestLoader().loadTestsFromTestCase(TestKalkulacka)
vysledek = unittest.TextTestRunner(verbosity=2).run(suite)
assert vysledek.wasSuccessful()
````

**Co si z cvičení 4 zapamatovat:** `setUp()` poskytne každému testu nový objekt. Testy tak na sobě nezávisí.

## Cvičení 5 — Kompletní úloha

Celý úkol je na vás od začátku do konce:

1. Vytvořte třídu `Kosik` (nákupní košík) s atributy:
   - `polozky` — seznam (list) slovníků `{"nazev": str, "cena": float, "pocet": int}`
2. Implementujte metody:
   - `pridej(nazev, cena, pocet=1)` — přidá položku do košíku
   - `celkova_cena()` — vrátí součet (cena × počet) všech položek
   - `pocet_polozek()` — vrátí celkový počet kusů v košíku
   - `odeber(nazev)` — odebere položku podle názvu; pokud neexistuje, vyvolá `ValueError`
3. Napište **alespoň 6 testů** pokrývajících:
   - Přidání položky a kontrola celkové ceny
   - Přidání více položek
   - Počet položek v košíku
   - Odebrání existující položky
   - Pokus o odebrání neexistující položky (výjimka)
   - Prázdný košík (celková cena = 0)

<details>
<summary>🔎 Očekávaný výstup (klikni pro zobrazení)</summary>

````python
import unittest

class Kosik:
    def __init__(self):
        self.polozky = []  # Každá položka je slovník s názvem, cenou a počtem.

    def pridej(self, nazev, cena, pocet=1):
        if cena < 0 or not isinstance(pocet, int) or pocet <= 0:
            raise ValueError("Cena nesmí být záporná a počet musí být kladné celé číslo")
        self.polozky.append({"nazev": nazev, "cena": float(cena), "pocet": pocet})

    def celkova_cena(self):
        # Pro každou položku násobíme cenu počtem kusů a výsledky sečteme.
        return sum(p["cena"] * p["pocet"] for p in self.polozky)

    def pocet_polozek(self):
        return sum(p["pocet"] for p in self.polozky)

    def odeber(self, nazev):
        for index, polozka in enumerate(self.polozky):
            if polozka["nazev"] == nazev:
                self.polozky.pop(index)  # Smaže celou nalezenou položku.
                return
        raise ValueError("Položka v košíku neexistuje")

class TestKosik(unittest.TestCase):
    def setUp(self):
        self.kosik = Kosik()  # Každý test začíná s prázdným košíkem.

    def test_pridej_polozku(self):
        self.kosik.pridej("Chléb", 30)
        self.assertEqual(len(self.kosik.polozky), 1)
        self.assertEqual(self.kosik.polozky[0]["nazev"], "Chléb")

    def test_celkova_cena(self):
        self.kosik.pridej("Voda", 15, 3)
        self.assertEqual(self.kosik.celkova_cena(), 45)

    def test_vice_polozek(self):
        self.kosik.pridej("Chléb", 30)
        self.kosik.pridej("Mléko", 25, 2)
        self.assertEqual(self.kosik.celkova_cena(), 80)

    def test_pocet_polozek(self):
        self.kosik.pridej("Voda", 10, 3)
        self.kosik.pridej("Chléb", 20, 2)
        self.assertEqual(self.kosik.pocet_polozek(), 5)

    def test_odeber_existujici(self):
        self.kosik.pridej("Voda", 10)
        self.kosik.odeber("Voda")
        self.assertEqual(self.kosik.pocet_polozek(), 0)

    def test_odeber_neexistujici(self):
        with self.assertRaises(ValueError):
            self.kosik.odeber("Neexistuje")

    def test_prazdny_kosik(self):
        self.assertEqual(self.kosik.celkova_cena(), 0)
        self.assertEqual(self.kosik.pocet_polozek(), 0)

suite = unittest.TestLoader().loadTestsFromTestCase(TestKosik)
vysledek = unittest.TextTestRunner(verbosity=2).run(suite)
assert vysledek.wasSuccessful()
````

**Co si z cvičení 5 zapamatovat:** Testy pokrývají přidání, odebrání, součet, počet kusů i neplatnou operaci.
